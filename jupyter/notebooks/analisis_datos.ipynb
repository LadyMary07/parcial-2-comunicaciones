# Verificación de Red Interna (Jupyter -> Joomla)
Este script de Python comprueba que Jupyter puede comunicarse exitosamente con el contenedor de Joomla a través de la red `frontend_net` de Docker.

```python
import urllib.request

url = 'http://joomla:80'
try:
    respuesta = urllib.request.urlopen(url)
    print(f'✅ Conexión exitosa a Joomla. Código HTTP recibido: {respuesta.getcode()}')
except Exception as e:
    print(f'❌ Error de conexión: {e}')
```
